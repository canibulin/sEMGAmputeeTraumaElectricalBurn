```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import featureExtraction4 as fea # feature extraction equations can be seen on  https://github.com/SebastianRestrepoA/EMG-pattern-recognition/tree/master
import segmentator as seg
from os import listdir
import re
```

```python
def find_csv_filenames( path_to_dir, suffix=".csv" ):
    filenames = listdir(path_to_dir)
    return [ filename for filename in filenames if filename.endswith( suffix ) ]
```

```python
pathDir='' #Path to there data is saved. In repository references to folder 1 sEMGLabeladData 
```

```python
features_names = ['VAR CH1', 'RMS CH1', 'IEMG CH1', 'MAV CH1', 'LOG CH1', 'WL CH1', 'ACC CH1', 'DASDV CH1', 'ZC CH1', 'WAMP CH1', 'MYOP CH1', "FR CH1",
                      "MNP CH1", "TP CH1","MNF CH1", "MDF CH1", "PKF CH1", "WENT CH1",
                      'VAR CH2', 'RMS CH2', 'IEMG CH2', 'MAV CH2', 'LOG CH2', 'WL CH2', 'ACC CH2', 'DASDV CH2', 'ZC CH2', 'WAMP CH2', 'MYOP CH2', "FR CH2",
                      "MNP CH2", "TP CH2","MNF CH2", "MDF CH2", "PKF CH2", "WENT CH2"
                     ]
```

```python
files=find_csv_filenames(pathDir)
arrBicepsTriceps = np.array([])
arrFlexorExtensor = np.array([])
for file in files:
    if file.find("Biceps") > 0 or file.find("Triceps") > 0:
        arrBicepsTriceps=np.append(arrBicepsTriceps,file)
    if file.find("Flexor") > 0 or file.find("Extensor") > 0:
        arrFlexorExtensor=np.append(arrFlexorExtensor,file)
    
```

```python
fs=512
tseg = 0.3  # segment duration (s)
nseg = int(tseg * fs)  # segment length (points)
noverlap = int(0.5 * nseg)  # overlap (points)
noffset = nseg - noverlap  # offset (points)
```

```python
ampFile=pd.read_csv("Amputados.csv")
```

```python
df = pd.DataFrame()
for file in arrBicepsTriceps:
    workFile=pd.read_csv(pathDir+"/"+file,low_memory=False)
    workFile.set_index('Samples') 
    res=fea.features_estimation(dataFrame=workFile, file_name=file, fs=fs, frame=nseg, step=noffset,
                                    plot=False,dataFrame2=ampFile)
    Joined= pd.concat([res[0].T, res[1],res[2]], axis=1)
    if df.empty:
        df=Joined
    else:
        df=pd.concat([df,Joined], axis=0)
    #except:
    #print("An error occured"+" : "+file) # An error occured
df.to_csv("BicepsTricepsProcessed6.csv")  
print("Finalizo!!")
```

```python
df = pd.DataFrame()
for file in arrFlexorExtensor:
    workFile=pd.read_csv(pathDir+"/"+file,low_memory=False)
    workFile.set_index('Samples') 
    #try:
    res=fea.features_estimation(dataFrame=workFile, file_name=file, fs=512, frame=nseg, step=noffset,
                                    plot=False,dataFrame2=ampFile)
    Joined= pd.concat([res[0].T, res[1],res[2]], axis=1)
    if df.empty:
        df=Joined
    else:
        df=pd.concat([df,Joined], axis=0)
    #except:
    #print("An error occured"+" : "+file) # An error occured
df.to_csv("FlexoExtension6.csv") 
print("Finalizo!!")
```
